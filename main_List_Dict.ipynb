Список Python-словарей — это комбинированная структура данных (коллекция), представляющая собой упорядоченную последовательность ([list]), где каждый отдельный элемент является неупорядоченным набором пар «ключ-значение» ([dict]).

В реальной разработке и базах данных такая структура является полным аналогом таблицы. Сам список list — это строки таблицы, а словари dict внутри него — это конкретные ячейки с данными в этих строках.

Сходства и различия list и dict


Критерий	Список (list)	Словарь (dict)
Синтаксис	Квадратные скобки: [...]	Фигурные скобки: {"ключ": значение}
Порядок элементов	Упорядочен. Элементы хранятся строго в том порядке, в котором их записали.	Упорядочен по времени добавления (начиная с Python 3.7+), но логически ориентирован на поиск по ключам.
Доступ к данным	По индексу (порядковому номеру): products[0]	По ключу (имени свойства): product["name"]
Дубликаты	Разрешены. Могут быть одинаковые элементы.	Ключи всегда уникальны. Значения могут повторяться.
Главная цель	Хранить перечень похожих объектов.	Хранить свойства (характеристики) одного объекта.

```python
# Это список (list), внутри которого лежат два словаря (dict)
ip_database = [
    {
        "unp": "190000001", 
        "vfio": "Иванов Иван Иванович", 
        "status": "Действующий"
    },
    {
        "unp": "190000002", 
        "vfio": "Петров Петр Петрович", 
        "status": "Ликвидирован"
    }
]

# 1. Как мы работаем со списком? Перебираем его циклом (строка за строкой):
for ip_card in ip_database:
    # 2. Как мы работаем со словарем? Достаем конкретное свойство по ключу:
    print(f"Предприниматель: {ip_card['vfio']} | {ip_card['unp']} | Статус: {ip_card['status']}")
```

```python
# Наш список с неизвестной и разной структурой словарей
unknown_list = [
    {"ФИО": "Иванов И.И.", "Статус": "Активен"},
    {"Компания": "ООО Вектор", "УНП": "190234", "Директор": "Петров П.П."},
    {"ID": 42, "Сообщение": "Успешно", "Теги": ["важно", "новое"]}
]

def print_unknown_structures(data_list):
    """Красиво выводит список словарей с неизвестной структурой."""
    
    # Внешний цикл: перебирает словари (строки нашей таблицы)
    for index, element in enumerate(data_list, start=1):
        print(f"--- Запись №{index} ---")
        
        # Проверяем, точно ли внутри лежит словарь
        if isinstance(element, dict):
            
            # Внутренний цикл: динамически достает каждый КЛЮЧ и его ЗНАЧЕНИЕ
            for key, value in element.items():
                print(f"  {key}: {value}")
                
        else:
            # На случай, если в список затесался не словарь, а обычная строка или число
            print(f"  Элемент не является словарем: {element}")
        
        print() # Пустая строка для отделения записей друг от друга

# Запуск функции
print_unknown_structures(unknown_list)
```

Как сделать вывод списка построчно, если структура словарей неизвестна?Если вы получили «тайный» список словарей и не знаете, какие ключи внутри них находятся, на помощь приходит динамический перебор. Словарь в Python умеет отдавать свои ключи и значения на лету с помощью метода .items().
Чтобы вывести такой список построчно и красиво, нужно использовать вложенный цикл (цикл в цикле).

```python
# Наш список с неизвестной и разной структурой словарей
unknown_list = [
    {"ФИО": "Иванов И.И.", "Статус": "Активен"},
    {"Компания": "ООО Вектор", "УНП": "190234", "Директор": "Петров П.П."},
    {"ID": 42, "Сообщение": "Успешно", "Теги": ["важно", "новое"]}
]

def print_unknown_structures(data_list):
    """Красиво выводит список словарей с неизвестной структурой."""
    
    # Внешний цикл: перебирает словари (строки нашей таблицы)
    for index, element in enumerate(data_list, start=1):
        print(f"--- Запись №{index} ---")
        
        # Проверяем, точно ли внутри лежит словарь
        if isinstance(element, dict):
            
            # Внутренний цикл: динамически достает каждый КЛЮЧ и его ЗНАЧЕНИЕ
            for key, value in element.items():
                print(f"  {key}: {value}")
                
        else:
            # На случай, если в список затесался не словарь, а обычная строка или число
            print(f"  Элемент не является словарем: {element}")
        
        print() # Пустая строка для отделения записей друг от друга

# Запуск функции
print_unknown_structures(unknown_list)
```
